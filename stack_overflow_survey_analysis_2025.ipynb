## 1. Підрахунок загальної кількості респондентів

```python
total_respondents = df.shape[0]
print("Загальна кількість респондентів:", total_respondents)
```

## 2. Аналіз повноти відповідей респондентів

```python
questions = list(set(df_schema['qname']).intersection(df.columns))

complete_respondents_count = df[questions].dropna().shape[0]

print("Респонденти, що відповіли на всі питання:", complete_respondents_count)
```

## 3. Статистичний аналіз досвіду респондентів

```python
work_exp = df['WorkExp'].dropna()
mean_exp = work_exp.mean()
median_exp = work_exp.median()
mode_exp = work_exp.mode().iloc[0]

print(f"Міри центральної тенденції - середнє: {mean_exp:.2f}, медіана: {median_exp}, мода: {mode_exp}")
```

## 4. Аналіз віддаленої роботи

```python
remote_workers_count = df['RemoteWork'].str.contains("Remote", na=False).sum()

print("Кількість респондентів, які працюють віддалено:", remote_workers_count)
```

## 5. Визначення популярності Python

```python
df['WorkedWithPython'] = df['LanguageHaveWorkedWith'].str.contains('Python', case=False, na=False)
python_users_count = df['WorkedWithPython'].sum()
python_percentage = python_users_count / len(df) * 100

print(f"Відсоток респондентів, що працюють з Python: {python_percentage:.1f}%")
```

## 6. Аналіз шляхів навчання програмуванню

```python
df['LearnedWithOnlineCourses'] = df['LearnCode'].str.contains('Online', case=False, na=False)
online_courses_count = df['LearnedWithOnlineCourses'].sum()

print("Респонденти, які навчалися через онлайн-курси:", online_courses_count)
```

## 7. Географічний аналіз компенсації Python-розробників

```python
python_devs = df[df['WorkedWithPython']].dropna(subset=['ConvertedCompYearly'])

py_by_country = python_devs.groupby('Country')['ConvertedCompYearly'].agg(
     mean_compensation='mean',
    median_compensation='median'
).sort_values('median_compensation', ascending=False)

py_by_country
```

## 8. Аналіз освіти найбільш оплачуваних спеціалістів

```python
top_paid_education = (
    df[['ResponseId', 'EdLevel', 'ConvertedCompYearly']]
    .dropna(subset=['ConvertedCompYearly'])
    .sort_values(by='ConvertedCompYearly', ascending=False)
    .head(5)
)

top_paid_education[['EdLevel']]
```

## 9. Аналіз популярності Python по віковим категоріям 

```python
age_python_stats = df.groupby('Age', dropna=False).agg(
    total_respondents=('WorkedWithPython', 'count'),
    python_users=('WorkedWithPython', 'sum')
).reset_index()

age_python_stats['python_percentage'] = (age_python_stats['python_users'] / age_python_stats['total_respondents'] * 100).round(2)

final_table = age_python_stats[['Age','python_percentage']].rename(
    columns={'Age':'Вікова категорія','python_percentage':'Відсоток Python-розробників'}
)

final_table
```

## 10. Аналіз індустрій серед високооплачуваних віддалених працівників

```python
salary_75 = df['ConvertedCompYearly'].quantile(0.75)
high_paid_remote = df[
    (df['ConvertedCompYearly'] > salary_75) &
    (df['RemoteWork'].str.contains("Remote", na=False))
]
industry_counts = high_paid_remote['Industry'].value_counts().reset_index()
industry_counts.columns = ['Індустрія', 'Кількість'] 

industry_counts
```

## Використання Python

```python
python_users = df['LanguageHaveWorkedWith'].str.contains("Python", na=False)

python_share = [
    python_users.sum(),
    len(df) - python_users.sum()
]

labels = ['Користувачі Python', 'Користувачі, які не користуються Python']
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
plt.pie(python_share, labels=labels, autopct='%1.1f%%')
plt.title("Використання Python серед розробників")
plt.show()
```

## Розподіл зарплат

```python
top_countries_salary = (
    df.groupby('Country')['ConvertedCompYearly']
    .median()
    .dropna()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
plt.figure(figsize=(10,5))
top_countries_salary.plot(kind='bar')
plt.title("10 найкращих країн за середньою зарплатою")
plt.xlabel("Країна")
plt.ylabel("Cередня зарплата (USD)")
plt.xticks(rotation=45)
plt.show()
```
